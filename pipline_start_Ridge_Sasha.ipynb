```python
# === Импорт библиотек ===
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_squared_error, r2_score

# === Загрузка данных ===
data = pd.read_csv("C:\\Users\\Aleks\\Downloads\\train_data.csv")



# Просмотр первых строк для проверки структуры
display(data.head())
print(len(data['region'].unique()))
```

```python
# === Разделение данных на признаки и целевую переменную ===
# === Удаляем строки с пропущенными значениями ===
data = data.dropna()

# === Удаляем координаты ===
X = data.drop(columns=['target', 'geo_lon', 'geo_lat'])
y = data['target']

# === Кодируем категориальный признак region ===
X = pd.get_dummies(X, columns=['region'], drop_first=True)

# === Теперь все признаки числовые ===
X = X.select_dtypes(include=['int64', 'float64', 'uint8'])

# === Базовый сплит данных (25% тест, seed = 42) ===
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)
```

```python
# === Обучение ===
model = Ridge(alpha=1.0, solver='auto', max_iter=1000)
model.fit(X_train, y_train)

# === Предсказание ===
y_pred = model.predict(X_test)

# === Метрики ===
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print(f"RMSE: {rmse:.4f}")
print(f"R²: {r2:.4f}")

# === Коэффициенты модели ===
print("\nКоэффициенты:", model.coef_)
print("Свободный член (intercept):", model.intercept_)
```

```python
from sklearn.linear_model import RidgeCV
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np

# === Список возможных значений α ===
alphas = [0.001, 0.01, 0.1, 1.0, 10.0, 100.0, 1000.0]

# === RidgeCV автоматически выберет лучшее значение α ===
model = RidgeCV(alphas=alphas, cv=5)  # store_cv_values убираем
model.fit(X_train, y_train)

# === Предсказание ===
y_pred = model.predict(X_test)

# === Метрики ===
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print(f"Лучшее значение alpha: {model.alpha_:.3f}")
print(f"RMSE: {rmse:.4f}")
print(f"R²: {r2:.4f}")
```

### Вывод и модификации данных

Для повышения интерпретируемости и устойчивости модели были выполнены следующие этапы предобработки данных:

1. **Удаление пропущенных значений**  
   Из исходного набора данных удалены строки с отсутствующими значениями (`NaN`), чтобы обеспечить корректную работу алгоритма Ridge-регрессии и исключить влияние некорректных наблюдений.

2. **Исключение координат (`geo_lon`, `geo_lat`)**  
   Географические координаты не несут прямой линейной зависимости с целевой переменной и могут вносить шум в модель.  
   Так как Ridge-регрессия строит линейную аппроксимацию, использование координат может искажать распределение весов и снижать обобщающую способность модели.  
   В связи с этим координаты были удалены.

3. **Кодирование категориального признака `region`**  
   Признак `region` содержал 76 уникальных значений, соответствующих регионам размещения банкоматов.  
   Для его включения в модель был применён **One-Hot Encoding** через функцию `pd.get_dummies(columns=['region'], drop_first=True)`.  
   Таким образом, каждый регион был представлен бинарным столбцом (0/1), что позволило сохранить региональные различия без нарушения линейности.  
   Использование параметра `drop_first=True` предотвратило проблему мультиколлинеарности.

4. **Удаление прочих категориальных признаков**  
   После кодирования `region` в модели остались только числовые признаки, включая сгенерированные бинарные индикаторы для каждого региона.

---

### Обучение модели Ridge-регрессии

Для обучения использовалась **Ridge-регрессия** — линейная модель с \( L_2 \)-регуляризацией, минимизирующая функционал:

$$
J(w) = \mathrm{MSE} + \alpha \lVert w \rVert^2
$$

Оптимальное значение коэффициента регуляризации подбиралось методом кросс-валидации через `RidgeCV`.  
Результаты подбора показали, что оптимальным значением является:

$$
\alpha = 1000
$$

- После удаления координат и добавления кодированных регионов качество модели **осталось стабильным**, при этом удалось повысить интерпретируемость коэффициентов и снизить влияние мультиколлинеарных признаков.  
- Оптимизация параметра регуляризации (α = 1000) привела к **незначительному улучшению метрики R²** и более сглаженному распределению весов, что говорит о снижении переобучения.  
- Учитывая невысокое значение R² (~0.41), модель описывает часть дисперсии целевой переменной, однако дальнейшее улучшение возможно за счёт добавления содержательных признаков.  
- Закодированный признак `region` позволил учесть влияние территориальных различий без использования координат, что соответствует предпосылкам линейной регрессии и улучшает устойчивость модели при обобщении.

Таким образом, проведённые модификации данных (удаление координат и кодирование `region`) обеспечили корректное применение Ridge-регрессии и повысили стабильность полученной модели.
